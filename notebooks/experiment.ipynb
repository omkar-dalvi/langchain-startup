```python
import asyncio

async def task(name, seconds):
    print(f"{name} started")

    await asyncio.sleep(seconds)

    print(f"{name} finished")
```

```python
import asyncio 
async def task(name, seconds): 
  print(f"{name} started") 
  await asyncio.sleep(seconds) 
  print(f"{name} finished") 

async def main(): 
  await asyncio.gather(
  task("A", 3),
  task("B", 3), 
  task("C", 3)
  )

await main()
```
